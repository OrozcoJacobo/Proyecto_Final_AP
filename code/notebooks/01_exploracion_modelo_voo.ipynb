## 4. Funciones auxiliares para conversión de formatos

El archivo CSV contiene algunas variables numéricas almacenadas como texto. Por ejemplo, los precios pueden incluir separadores de miles, el volumen puede estar abreviado con letras como `K`, `M` o `B`, y el cambio porcentual incluye el símbolo `%`.

Para evitar repetir lógica de limpieza en varias partes del notebook, se definen funciones auxiliares. Estas funciones se documentan con un formato compatible con MkDocs, especificando su propósito, parámetros, retorno y posibles consideraciones.

```python
def convertir_precio_a_float(valor_original):
    """
    Convierte un valor de precio del dataset a número decimal.

    Esta función limpia valores de precio que pueden venir como texto,
    eliminando separadores de miles y espacios antes de convertirlos
    a tipo `float`.

    Args:
        valor_original: Valor original de una columna de precio.

    Returns:
        Valor numérico convertido a `float`. Si el valor original es nulo,
        retorna `np.nan`.
    """
    if pd.isna(valor_original):
        return np.nan

    valor_limpio = str(valor_original).replace(",", "").strip()

    return float(valor_limpio)


def convertir_porcentaje_a_decimal(valor_original):
    """
    Convierte un porcentaje textual a proporción decimal.

    Por ejemplo, el valor `"1.25%"` se transforma en `0.0125`.

    Args:
        valor_original: Valor original de la columna porcentual.

    Returns:
        Valor porcentual convertido a proporción decimal. Si el valor
        original es nulo, retorna `np.nan`.
    """
    if pd.isna(valor_original):
        return np.nan

    valor_limpio = str(valor_original).replace("%", "").replace(",", "").strip()

    return float(valor_limpio) / 100


def convertir_volumen_a_float(valor_original):
    """
    Convierte el volumen transado a valor numérico.

    El dataset puede representar el volumen con abreviaciones como `K`
    para miles, `M` para millones o `B` para miles de millones.

    Args:
        valor_original: Valor original de la columna de volumen.

    Returns:
        Volumen convertido a `float`. Si el valor original es nulo o no
        contiene información válida, retorna `np.nan`.
    """
    if pd.isna(valor_original):
        return np.nan

    valor_texto = str(valor_original).replace(",", "").strip().upper()

    if valor_texto in {"", "-", "NAN"}:
        return np.nan

    multiplicador = 1.0

    if valor_texto.endswith("K"):
        multiplicador = 1_000
        valor_texto = valor_texto[:-1]
    elif valor_texto.endswith("M"):
        multiplicador = 1_000_000
        valor_texto = valor_texto[:-1]
    elif valor_texto.endswith("B"):
        multiplicador = 1_000_000_000
        valor_texto = valor_texto[:-1]

    return float(valor_texto) * multiplicador
```

## 11. Tratamiento de registros sin operación

Esta decisión es metodológicamente más sólida que imputar el volumen, porque imputar implicaría crear artificialmente actividad de mercado en días donde no hubo negociación reportada. Para un modelo financiero, este tipo de registros puede introducir ruido y afectar la construcción de variables basadas en retornos, volatilidad o volumen.

```python
registros_sin_operacion = (
    datos_voo["volumen"].isna()
    & (datos_voo["precio_cierre"] == datos_voo["precio_apertura"])
    & (datos_voo["precio_cierre"] == datos_voo["precio_maximo"])
    & (datos_voo["precio_cierre"] == datos_voo["precio_minimo"])
    & (datos_voo["cambio_porcentual"] == 0)
)

datos_sin_operacion = datos_voo.loc[registros_sin_operacion].copy()
datos_sin_operacion["dia_semana"] = datos_sin_operacion["fecha"].dt.day_name()

print(f"Registros con volumen nulo: {datos_voo['volumen'].isna().sum()}")
print(f"Registros identificados como sin operación: {registros_sin_operacion.sum()}")

display(datos_sin_operacion)
```

## 12. Definición de la variable objetivo

Para convertir el problema financiero en una tarea de aprendizaje supervisado, se define una variable objetivo binaria. El objetivo no es predecir exactamente el precio futuro del ETF, sino estimar si existe una oportunidad de inversión en un horizonte temporal determinado.

En este proyecto se entiende como oportunidad de inversión un escenario en el que el precio futuro del ETF supera el precio actual en una proporción mínima definida. Esta condición evita que el modelo clasifique como oportunidad cualquier aumento marginal de precio, lo cual sería poco útil para un usuario final.

Se evaluarán tres horizontes temporales:

- 4 días hábiles: representa una señal de muy corto plazo.
- 15 días hábiles: representa una señal de corto plazo.
- 20 días hábiles: aproxima un horizonte mensual de mercado.

Para cada horizonte se calcula el retorno futuro como:

\[
retorno\ futuro = \frac{precio\ futuro - precio\ actual}{precio\ actual}
\]

Luego, se asigna la clase positiva cuando el retorno futuro supera un umbral mínimo. Esta formulación permite entrenar modelos de clasificación binaria y posteriormente construir una capa de decisión interpretable para la aplicación.

```python
def calcular_retorno_futuro(datos, horizonte_dias):
    """
    Calcula el retorno futuro del precio de cierre para un horizonte dado.

    El retorno futuro compara el precio de cierre actual con el precio de
    cierre observado después de una cantidad determinada de días hábiles.

    Args:
        datos: DataFrame que contiene la columna `precio_cierre`.
        horizonte_dias: Número de días hacia adelante usados para calcular
            el retorno futuro.

    Returns:
        Serie de pandas con el retorno futuro para cada observación.
        Las últimas observaciones quedan como `NaN`, porque no existe
        información futura suficiente para calcular el retorno.
    """
    precio_futuro = datos["precio_cierre"].shift(-horizonte_dias)
    retorno_futuro = (precio_futuro - datos["precio_cierre"]) / datos["precio_cierre"]

    return retorno_futuro


def crear_variable_objetivo(datos, horizonte_dias, umbral_retorno):
    """
    Crea una variable objetivo binaria para detectar oportunidades de inversión.

    La clase positiva se asigna cuando el retorno futuro del ETF es mayor o
    igual al umbral definido para el horizonte temporal analizado.

    Args:
        datos: DataFrame limpio con la columna `precio_cierre`.
        horizonte_dias: Número de días hábiles hacia adelante.
        umbral_retorno: Retorno mínimo requerido para considerar que existe
            una oportunidad de inversión.

    Returns:
        DataFrame con dos nuevas columnas:
        `retorno_futuro_{horizonte_dias}d` y
        `objetivo_oportunidad_{horizonte_dias}d`.
    """
    datos_con_objetivo = datos.copy()

    columna_retorno = f"retorno_futuro_{horizonte_dias}d"
    columna_objetivo = f"objetivo_oportunidad_{horizonte_dias}d"

    datos_con_objetivo[columna_retorno] = calcular_retorno_futuro(
        datos=datos_con_objetivo,
        horizonte_dias=horizonte_dias,
    )

    datos_con_objetivo[columna_objetivo] = np.where(
        datos_con_objetivo[columna_retorno] >= umbral_retorno,
        1,
        0,
    )

    datos_con_objetivo.loc[
        datos_con_objetivo[columna_retorno].isna(),
        columna_objetivo,
    ] = np.nan

    return datos_con_objetivo
```

## 13. Construcción de objetivos por horizonte

Se crean variables objetivo para tres horizontes temporales. Los umbrales no son iguales para todos los horizontes, porque no es razonable exigir el mismo retorno mínimo en 4 días que en 20 días.

Para el horizonte de 4 días se utiliza un umbral menor, mientras que para horizontes más largos se exige un retorno mayor. Esta decisión busca que la clase positiva represente una oportunidad relevante y no simplemente cualquier variación mínima del precio.

```python
configuracion_objetivos = {
    4: 0.005,
    15: 0.015,
    20: 0.020,
}

datos_modelado = datos_voo.copy()

for horizonte_dias, umbral_retorno in configuracion_objetivos.items():
    datos_modelado = crear_variable_objetivo(
        datos=datos_modelado,
        horizonte_dias=horizonte_dias,
        umbral_retorno=umbral_retorno,
    )

columnas_objetivo = [
    columna
    for columna in datos_modelado.columns
    if columna.startswith("objetivo_oportunidad")
]

columnas_retorno_futuro = [
    columna
    for columna in datos_modelado.columns
    if columna.startswith("retorno_futuro")
]

display(datos_modelado[
    ["fecha", "precio_cierre"] + columnas_retorno_futuro + columnas_objetivo
].head())

display(datos_modelado[
    ["fecha", "precio_cierre"] + columnas_retorno_futuro + columnas_objetivo
].tail(25))
```

## 14. Revisión del balance de clases

Después de construir las variables objetivo, se revisa la proporción de clases para cada horizonte. Este paso es importante porque un dataset muy desbalanceado puede producir modelos aparentemente buenos, pero poco útiles.

Por ejemplo, si la mayoría de observaciones pertenecen a la clase positiva, un modelo que siempre prediga oportunidad podría obtener una exactitud alta sin realmente aprender patrones relevantes. Por esta razón se revisan tanto los conteos como los porcentajes de cada clase.

```python
resumen_balance_clases = []

for horizonte_dias in configuracion_objetivos:
    columna_objetivo = f"objetivo_oportunidad_{horizonte_dias}d"
    datos_validos_objetivo = datos_modelado.dropna(subset=[columna_objetivo]).copy()

    conteos_clase = datos_validos_objetivo[columna_objetivo].value_counts().sort_index()
    porcentajes_clase = datos_validos_objetivo[columna_objetivo].value_counts(normalize=True).sort_index()

    resumen_balance_clases.append({
        "horizonte_dias": horizonte_dias,
        "umbral_retorno": configuracion_objetivos[horizonte_dias],
        "total_observaciones": datos_validos_objetivo.shape[0],
        "clase_0_conteo": int(conteos_clase.get(0.0, 0)),
        "clase_1_conteo": int(conteos_clase.get(1.0, 0)),
        "clase_0_porcentaje": porcentajes_clase.get(0.0, 0),
        "clase_1_porcentaje": porcentajes_clase.get(1.0, 0),
    })

balance_clases = pd.DataFrame(resumen_balance_clases)

display(balance_clases)
```

```python
plt.figure(figsize=(10, 5))

posiciones_horizonte = np.arange(len(balance_clases))
ancho_barra = 0.35

plt.bar(
    posiciones_horizonte - ancho_barra / 2,
    balance_clases["clase_0_porcentaje"],
    width=ancho_barra,
    label="Clase 0: no oportunidad",
)

plt.bar(
    posiciones_horizonte + ancho_barra / 2,
    balance_clases["clase_1_porcentaje"],
    width=ancho_barra,
    label="Clase 1: oportunidad",
)

plt.xticks(
    posiciones_horizonte,
    balance_clases["horizonte_dias"].astype(str) + " días",
)

plt.title("Balance de clases por horizonte de predicción")
plt.xlabel("Horizonte")
plt.ylabel("Proporción")
plt.legend()
plt.grid(True, axis="y")
plt.show()
```

La revisión del balance de clases muestra que los tres horizontes definidos generan una distribución relativamente equilibrada entre la clase positiva y la clase negativa.

Para el horizonte de 4 días, la clase positiva representa aproximadamente el 45,5% de las observaciones. Para 15 días, representa aproximadamente el 45,1%, y para 20 días representa aproximadamente el 42,7%. Esto indica que los umbrales definidos no producen un problema extremadamente desbalanceado.

Esta observación es importante porque evita que el modelo pueda obtener buenos resultados simplemente prediciendo siempre la clase mayoritaria. Sin embargo, esta revisión todavía es global. En etapas posteriores será necesario revisar el balance de clases en los conjuntos de entrenamiento, validación y prueba, respetando el orden temporal de los datos.

Con base en estos resultados, se mantienen los tres horizontes propuestos:

- 4 días con umbral de retorno mínimo de 0,5%.
- 15 días con umbral de retorno mínimo de 1,5%.
- 20 días con umbral de retorno mínimo de 2,0%.

Estos horizontes permitirán que la aplicación final genere una recomendación más rica que una simple respuesta de comprar o no comprar, ya que podrá diferenciar entre señales de muy corto plazo, corto plazo y horizonte mensual.

## 15. Ingeniería de características

Los modelos de aprendizaje de máquina no se entrenarán únicamente con las variables originales del dataset. Aunque las columnas de precio y volumen contienen información relevante, utilizar solo sus valores absolutos puede ser insuficiente para capturar el comportamiento financiero del ETF.

Por esta razón, se construyen variables derivadas que resumen el comportamiento reciente del activo. Estas variables buscan representar conceptos como retorno, volatilidad, momentum, distancia frente a máximos o mínimos recientes y cambios en volumen.

Todas las variables predictoras se construyen utilizando únicamente información disponible hasta la fecha actual de cada observación. Esto es fundamental para evitar fuga de información, es decir, evitar que el modelo aprenda usando datos del futuro.

```python
def agregar_variables_de_retorno(datos):
    """
    Agrega variables basadas en retornos históricos del precio.

    Estas variables describen cambios porcentuales recientes del precio de
    cierre. Se utilizan retornos simples y retornos logarítmicos porque son
    más informativos que el precio absoluto cuando se trabaja con series
    financieras.

    Args:
        datos: DataFrame con la columna `precio_cierre`.

    Returns:
        DataFrame con nuevas columnas de retornos históricos.
    """
    datos_features = datos.copy()

    datos_features["retorno_diario"] = datos_features["precio_cierre"].pct_change()
    datos_features["retorno_logaritmico"] = np.log(
        datos_features["precio_cierre"] / datos_features["precio_cierre"].shift(1)
    )

    datos_features["retorno_3d"] = datos_features["precio_cierre"].pct_change(3)
    datos_features["retorno_5d"] = datos_features["precio_cierre"].pct_change(5)
    datos_features["retorno_10d"] = datos_features["precio_cierre"].pct_change(10)
    datos_features["retorno_20d"] = datos_features["precio_cierre"].pct_change(20)

    return datos_features


def agregar_variables_de_rango(datos):
    """
    Agrega variables relacionadas con el rango diario de negociación.

    Estas variables describen la diferencia entre precio máximo, mínimo,
    apertura y cierre. Son útiles para aproximar volatilidad diaria y presión
    compradora o vendedora dentro de cada jornada.

    Args:
        datos: DataFrame con columnas de precios OHLC.

    Returns:
        DataFrame con variables derivadas del rango diario.
    """
    datos_features = datos.copy()

    datos_features["rango_diario"] = (
        datos_features["precio_maximo"] - datos_features["precio_minimo"]
    ) / datos_features["precio_cierre"]

    datos_features["cambio_apertura_cierre"] = (
        datos_features["precio_cierre"] - datos_features["precio_apertura"]
    ) / datos_features["precio_apertura"]

    datos_features["posicion_cierre_rango"] = (
        datos_features["precio_cierre"] - datos_features["precio_minimo"]
    ) / (
        datos_features["precio_maximo"] - datos_features["precio_minimo"]
    )

    datos_features["posicion_cierre_rango"] = datos_features[
        "posicion_cierre_rango"
    ].replace([np.inf, -np.inf], np.nan)

    return datos_features


def agregar_medias_moviles(datos, ventanas):
    """
    Agrega medias móviles y distancia del precio frente a esas medias.

    Las medias móviles resumen la tendencia reciente del precio. La distancia
    entre el precio actual y su media móvil permite identificar si el activo
    está relativamente extendido o rezagado frente a su comportamiento reciente.

    Args:
        datos: DataFrame con la columna `precio_cierre`.
        ventanas: Lista de ventanas temporales para calcular medias móviles.

    Returns:
        DataFrame con medias móviles y distancias relativas.
    """
    datos_features = datos.copy()

    for ventana in ventanas:
        columna_media = f"media_movil_{ventana}d"
        columna_distancia = f"distancia_media_movil_{ventana}d"

        datos_features[columna_media] = (
            datos_features["precio_cierre"]
            .rolling(window=ventana)
            .mean()
        )

        datos_features[columna_distancia] = (
            datos_features["precio_cierre"] - datos_features[columna_media]
        ) / datos_features[columna_media]

    return datos_features


def agregar_variables_de_volatilidad(datos, ventanas):
    """
    Agrega variables de volatilidad histórica.

    La volatilidad se calcula como la desviación estándar de los retornos
    diarios en diferentes ventanas temporales. Esta información permite
    representar periodos de mayor o menor incertidumbre en el mercado.

    Args:
        datos: DataFrame con la columna `retorno_diario`.
        ventanas: Lista de ventanas temporales para calcular volatilidad.

    Returns:
        DataFrame con variables de volatilidad histórica.
    """
    datos_features = datos.copy()

    for ventana in ventanas:
        datos_features[f"volatilidad_{ventana}d"] = (
            datos_features["retorno_diario"]
            .rolling(window=ventana)
            .std()
        )

    return datos_features


def agregar_variables_de_momentum(datos, ventanas):
    """
    Agrega variables de momentum del precio.

    El momentum mide el cambio acumulado del precio en ventanas recientes.
    Puede ayudar a identificar si el ETF viene en una tendencia positiva o
    negativa antes de la fecha evaluada.

    Args:
        datos: DataFrame con la columna `precio_cierre`.
        ventanas: Lista de horizontes pasados para calcular momentum.

    Returns:
        DataFrame con variables de momentum.
    """
    datos_features = datos.copy()

    for ventana in ventanas:
        datos_features[f"momentum_{ventana}d"] = (
            datos_features["precio_cierre"] / datos_features["precio_cierre"].shift(ventana)
        ) - 1

    return datos_features


def agregar_distancia_a_extremos(datos, ventanas):
    """
    Agrega variables de distancia frente a máximos y mínimos recientes.

    Estas variables permiten medir si el precio actual está cerca de un máximo
    reciente o de un mínimo reciente. Esto es relevante para el proyecto porque
    la aplicación busca apoyar la identificación de posibles momentos de entrada.

    Args:
        datos: DataFrame con la columna `precio_cierre`.
        ventanas: Lista de ventanas temporales para calcular máximos y mínimos.

    Returns:
        DataFrame con distancias relativas a máximos y mínimos recientes.
    """
    datos_features = datos.copy()

    for ventana in ventanas:
        maximo_reciente = datos_features["precio_cierre"].rolling(window=ventana).max()
        minimo_reciente = datos_features["precio_cierre"].rolling(window=ventana).min()

        datos_features[f"distancia_maximo_{ventana}d"] = (
            datos_features["precio_cierre"] - maximo_reciente
        ) / maximo_reciente

        datos_features[f"distancia_minimo_{ventana}d"] = (
            datos_features["precio_cierre"] - minimo_reciente
        ) / minimo_reciente

    return datos_features


def agregar_variables_de_volumen(datos, ventanas):
    """
    Agrega variables derivadas del volumen transado.

    El volumen puede reflejar cambios en la actividad del mercado. En lugar de
    usar únicamente el volumen absoluto, se crean variables relativas frente a
    su comportamiento reciente.

    Args:
        datos: DataFrame con la columna `volumen`.
        ventanas: Lista de ventanas temporales para calcular medias de volumen.

    Returns:
        DataFrame con variables derivadas del volumen.
    """
    datos_features = datos.copy()

    datos_features["cambio_volumen"] = datos_features["volumen"].pct_change()
    datos_features["cambio_volumen"] = datos_features["cambio_volumen"].replace(
        [np.inf, -np.inf],
        np.nan,
    )

    for ventana in ventanas:
        columna_media_volumen = f"media_volumen_{ventana}d"
        columna_ratio_volumen = f"ratio_volumen_{ventana}d"

        datos_features[columna_media_volumen] = (
            datos_features["volumen"]
            .rolling(window=ventana)
            .mean()
        )

        datos_features[columna_ratio_volumen] = (
            datos_features["volumen"] / datos_features[columna_media_volumen]
        )

    return datos_features


def construir_variables_predictoras(datos):
    """
    Construye el conjunto completo de variables predictoras.

    Esta función aplica de forma ordenada las transformaciones necesarias para
    enriquecer el dataset con variables financieras derivadas. Todas las
    variables se construyen usando información histórica o contemporánea
    disponible para cada fecha.

    Args:
        datos: DataFrame limpio con precios, volumen y variables objetivo.

    Returns:
        DataFrame enriquecido con variables predictoras para modelado.
    """
    ventanas_cortas = [5, 10, 20]
    ventanas_largas = [20, 50, 100]

    datos_features = datos.copy()

    datos_features = agregar_variables_de_retorno(datos_features)
    datos_features = agregar_variables_de_rango(datos_features)
    datos_features = agregar_medias_moviles(datos_features, ventanas_largas)
    datos_features = agregar_variables_de_volatilidad(datos_features, ventanas_cortas)
    datos_features = agregar_variables_de_momentum(datos_features, ventanas_cortas)
    datos_features = agregar_distancia_a_extremos(datos_features, ventanas_largas)
    datos_features = agregar_variables_de_volumen(datos_features, ventanas_cortas)

    return datos_features
```

Adicionalmente, no se usarán directamente los precios absolutos ni el volumen absoluto como variables principales del modelo. En su lugar, se priorizan variables relativas como retornos, distancias porcentuales, volatilidades, momentum y ratios de volumen. Esto hace que el modelo dependa menos del nivel histórico del precio y más del comportamiento reciente del activo.

## 19. Preparación del dataset de modelado por horizonte

Como se construyeron tres variables objetivo diferentes, se preparará un dataset de modelado independiente para cada horizonte. Esto permite entrenar y evaluar modelos específicos para 4, 15 y 20 días.

Para cada horizonte se eliminan únicamente las filas que tienen valores faltantes en las variables predictoras o en la variable objetivo correspondiente. No se eliminan filas por valores faltantes de objetivos de otros horizontes, ya que esos objetivos no son necesarios para entrenar el modelo del horizonte actual.

Esta decisión evita perder observaciones innecesariamente y mantiene la preparación metodológicamente alineada con cada problema de clasificación.

```python
def preparar_dataset_por_horizonte(datos, columnas_entrada, horizonte_dias):
    """
    Prepara el dataset de modelado para un horizonte específico.

    Esta función selecciona las variables predictoras y la variable objetivo
    asociada al horizonte indicado. Luego elimina filas con valores faltantes
    únicamente en las columnas necesarias para ese modelo.

    Args:
        datos: DataFrame con variables predictoras y objetivos.
        columnas_entrada: Lista de columnas que serán usadas como variables
            predictoras.
        horizonte_dias: Horizonte temporal del objetivo, expresado en días
            hábiles.

    Returns:
        DataFrame limpio para modelado, con la columna `fecha`, las variables
        predictoras y la variable objetivo correspondiente.
    """
    columna_objetivo = f"objetivo_oportunidad_{horizonte_dias}d"

    columnas_modelo = ["fecha"] + columnas_entrada + [columna_objetivo]

    datos_horizonte = datos[columnas_modelo].copy()
    datos_horizonte = datos_horizonte.dropna().reset_index(drop=True)

    datos_horizonte[columna_objetivo] = datos_horizonte[columna_objetivo].astype(int)

    return datos_horizonte
```

```python
datasets_por_horizonte = {}

for horizonte_dias in configuracion_objetivos:
    datasets_por_horizonte[horizonte_dias] = preparar_dataset_por_horizonte(
        datos=datos_features,
        columnas_entrada=columnas_predictoras,
        horizonte_dias=horizonte_dias,
    )

    print(f"Horizonte {horizonte_dias} días")
    print(f"Filas disponibles: {datasets_por_horizonte[horizonte_dias].shape[0]}")
    print(f"Columnas disponibles: {datasets_por_horizonte[horizonte_dias].shape[1]}")
    print("-" * 40)
```

## 20. División temporal en entrenamiento, validación y prueba

Debido a que el dataset corresponde a una serie temporal financiera, no se debe usar una división aleatoria de los datos. Una división aleatoria mezclaría observaciones antiguas y recientes, permitiendo que el modelo aprenda indirectamente información del futuro.

Por esta razón, se utiliza una división cronológica:

- 70% inicial para entrenamiento.
- 15% siguiente para validación.
- 15% final para prueba.

El conjunto de entrenamiento se usa para ajustar los modelos, el conjunto de validación para comparar configuraciones e hiperparámetros, y el conjunto de prueba para evaluar el desempeño final sobre datos más recientes no utilizados durante la selección del modelo.

```python
def dividir_dataset_temporal(datos, columna_objetivo, proporcion_entrenamiento=0.70, proporcion_validacion=0.15):
    """
    Divide un dataset en entrenamiento, validación y prueba respetando el orden temporal.

    Esta función evita mezclar observaciones futuras con pasadas. La división se
    realiza de forma cronológica, usando las primeras observaciones para
    entrenamiento, las siguientes para validación y las últimas para prueba.

    Args:
        datos: DataFrame limpio para modelado.
        columna_objetivo: Nombre de la columna objetivo.
        proporcion_entrenamiento: Proporción inicial del dataset usada para
            entrenamiento.
        proporcion_validacion: Proporción siguiente usada para validación.

    Returns:
        Diccionario con los conjuntos `x_entrenamiento`, `y_entrenamiento`,
        `x_validacion`, `y_validacion`, `x_prueba`, `y_prueba`, junto con las
        fechas correspondientes de cada partición.
    """
    total_registros = datos.shape[0]

    indice_fin_entrenamiento = int(total_registros * proporcion_entrenamiento)
    indice_fin_validacion = int(total_registros * (proporcion_entrenamiento + proporcion_validacion))

    datos_entrenamiento = datos.iloc[:indice_fin_entrenamiento].copy()
    datos_validacion = datos.iloc[indice_fin_entrenamiento:indice_fin_validacion].copy()
    datos_prueba = datos.iloc[indice_fin_validacion:].copy()

    x_entrenamiento = datos_entrenamiento[columnas_predictoras]
    y_entrenamiento = datos_entrenamiento[columna_objetivo]

    x_validacion = datos_validacion[columnas_predictoras]
    y_validacion = datos_validacion[columna_objetivo]

    x_prueba = datos_prueba[columnas_predictoras]
    y_prueba = datos_prueba[columna_objetivo]

    return {
        "datos_entrenamiento": datos_entrenamiento,
        "datos_validacion": datos_validacion,
        "datos_prueba": datos_prueba,
        "x_entrenamiento": x_entrenamiento,
        "y_entrenamiento": y_entrenamiento,
        "x_validacion": x_validacion,
        "y_validacion": y_validacion,
        "x_prueba": x_prueba,
        "y_prueba": y_prueba,
    }
```

```python
particiones_por_horizonte = {}

for horizonte_dias, datos_horizonte in datasets_por_horizonte.items():
    columna_objetivo = f"objetivo_oportunidad_{horizonte_dias}d"

    particiones_por_horizonte[horizonte_dias] = dividir_dataset_temporal(
        datos=datos_horizonte,
        columna_objetivo=columna_objetivo,
    )
```

## 21. Revisión de las particiones temporales

Después de dividir los datos, se revisan los rangos de fechas y el balance de clases en cada partición. Esta validación es importante porque, aunque el dataset completo tenga un balance razonable, la distribución puede cambiar entre periodos históricos.

En mercados financieros, este punto es especialmente relevante porque el comportamiento del activo puede variar entre ciclos de crecimiento, crisis, recuperación o alta volatilidad.

```python
resumen_particiones = []

for horizonte_dias, particiones in particiones_por_horizonte.items():
    columna_objetivo = f"objetivo_oportunidad_{horizonte_dias}d"

    for nombre_particion in ["datos_entrenamiento", "datos_validacion", "datos_prueba"]:
        datos_particion = particiones[nombre_particion]

        conteos_clase = datos_particion[columna_objetivo].value_counts().sort_index()
        porcentajes_clase = datos_particion[columna_objetivo].value_counts(normalize=True).sort_index()

        resumen_particiones.append({
            "horizonte_dias": horizonte_dias,
            "particion": nombre_particion.replace("datos_", ""),
            "fecha_inicio": datos_particion["fecha"].min(),
            "fecha_fin": datos_particion["fecha"].max(),
            "cantidad_registros": datos_particion.shape[0],
            "clase_0_conteo": int(conteos_clase.get(0, 0)),
            "clase_1_conteo": int(conteos_clase.get(1, 0)),
            "clase_0_porcentaje": porcentajes_clase.get(0, 0),
            "clase_1_porcentaje": porcentajes_clase.get(1, 0),
        })

resumen_particiones = pd.DataFrame(resumen_particiones)

display(resumen_particiones)
```

## Interpretación de la preparación para modelado

## 22. Entrenamiento inicial de modelos

En esta sección se entrenan diferentes modelos de clasificación supervisada para cada horizonte temporal. Se incluyen modelos de distinta naturaleza para cumplir con el objetivo metodológico del proyecto:

- Un modelo ingenuo de referencia, que siempre predice la clase mayoritaria.
- Regresión logística, como modelo lineal base.
- Random Forest, como método de ensamble basado en árboles.
- Gradient Boosting, como método de ensamble secuencial.
- Red neuronal multicapa, como modelo no lineal basado en redes neuronales artificiales.

El modelo ingenuo no se considera una solución final, pero sirve como línea base. Si los modelos de aprendizaje de máquina no superan claramente esta referencia, entonces no habría evidencia suficiente de que están aprendiendo patrones útiles.

En esta primera etapa se usan configuraciones iniciales razonables. Posteriormente se realizará exploración de hiperparámetros sobre los modelos más prometedores.

```python
from sklearn.dummy import DummyClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
)
```

```python
def crear_modelos_candidatos():
    """
    Crea los modelos candidatos para el entrenamiento inicial.

    Los modelos se definen en una función para garantizar que cada horizonte
    reciba instancias nuevas e independientes. Esto evita reutilizar modelos
    previamente ajustados en otros horizontes.

    Returns:
        Diccionario donde las llaves son los nombres de los modelos y los
        valores son estimadores compatibles con scikit-learn.
    """
    modelos_candidatos = {
        "dummy_mayoria": DummyClassifier(strategy="most_frequent"),
        "regresion_logistica": Pipeline(
            steps=[
                ("escalador", StandardScaler()),
                (
                    "modelo",
                    LogisticRegression(
                        max_iter=2000,
                        random_state=42,
                    ),
                ),
            ]
        ),
        "random_forest": RandomForestClassifier(
            n_estimators=300,
            max_depth=8,
            min_samples_leaf=10,
            random_state=42,
            n_jobs=-1,
        ),
        "gradient_boosting": GradientBoostingClassifier(
            n_estimators=150,
            learning_rate=0.05,
            max_depth=3,
            random_state=42,
        ),
        "red_neuronal_mlp": Pipeline(
            steps=[
                ("escalador", StandardScaler()),
                (
                    "modelo",
                    MLPClassifier(
                        hidden_layer_sizes=(32, 16),
                        activation="relu",
                        alpha=0.001,
                        learning_rate_init=0.001,
                        max_iter=1000,
                        early_stopping=True,
                        random_state=42,
                    ),
                ),
            ]
        ),
    }

    return modelos_candidatos
```

```python
def evaluar_modelo_clasificacion(modelo, x_datos, y_real):
    """
    Evalúa un modelo de clasificación binaria.

    La función calcula métricas estándar de clasificación usando las
    predicciones discretas y las probabilidades estimadas para la clase
    positiva.

    Args:
        modelo: Modelo de clasificación previamente entrenado.
        x_datos: Variables predictoras del conjunto a evaluar.
        y_real: Etiquetas reales del conjunto a evaluar.

    Returns:
        Diccionario con accuracy, precision, recall, f1 y roc_auc.
    """
    y_predicho = modelo.predict(x_datos)

    if hasattr(modelo, "predict_proba"):
        probabilidades_clase_positiva = modelo.predict_proba(x_datos)[:, 1]
    else:
        probabilidades_clase_positiva = y_predicho

    metricas = {
        "accuracy": accuracy_score(y_real, y_predicho),
        "precision": precision_score(y_real, y_predicho, zero_division=0),
        "recall": recall_score(y_real, y_predicho, zero_division=0),
        "f1": f1_score(y_real, y_predicho, zero_division=0),
        "roc_auc": roc_auc_score(y_real, probabilidades_clase_positiva),
    }

    return metricas
```

```python
resultados_entrenamiento_inicial = []
modelos_entrenados_iniciales = {}

for horizonte_dias, particiones in particiones_por_horizonte.items():
    modelos_entrenados_iniciales[horizonte_dias] = {}

    x_entrenamiento = particiones["x_entrenamiento"]
    y_entrenamiento = particiones["y_entrenamiento"]
    x_validacion = particiones["x_validacion"]
    y_validacion = particiones["y_validacion"]

    modelos_candidatos = crear_modelos_candidatos()

    for nombre_modelo, modelo in modelos_candidatos.items():
        modelo.fit(x_entrenamiento, y_entrenamiento)

        metricas_entrenamiento = evaluar_modelo_clasificacion(
            modelo=modelo,
            x_datos=x_entrenamiento,
            y_real=y_entrenamiento,
        )

        metricas_validacion = evaluar_modelo_clasificacion(
            modelo=modelo,
            x_datos=x_validacion,
            y_real=y_validacion,
        )

        modelos_entrenados_iniciales[horizonte_dias][nombre_modelo] = modelo

        resultados_entrenamiento_inicial.append({
            "horizonte_dias": horizonte_dias,
            "modelo": nombre_modelo,
            "accuracy_entrenamiento": metricas_entrenamiento["accuracy"],
            "precision_entrenamiento": metricas_entrenamiento["precision"],
            "recall_entrenamiento": metricas_entrenamiento["recall"],
            "f1_entrenamiento": metricas_entrenamiento["f1"],
            "roc_auc_entrenamiento": metricas_entrenamiento["roc_auc"],
            "accuracy_validacion": metricas_validacion["accuracy"],
            "precision_validacion": metricas_validacion["precision"],
            "recall_validacion": metricas_validacion["recall"],
            "f1_validacion": metricas_validacion["f1"],
            "roc_auc_validacion": metricas_validacion["roc_auc"],
        })

resultados_entrenamiento_inicial = pd.DataFrame(resultados_entrenamiento_inicial)

display(
    resultados_entrenamiento_inicial.sort_values(
        by=["horizonte_dias", "f1_validacion"],
        ascending=[True, False],
    )
)
```

```python
plt.figure(figsize=(12, 6))

for horizonte_dias in sorted(resultados_entrenamiento_inicial["horizonte_dias"].unique()):
    resultados_horizonte = resultados_entrenamiento_inicial[
        resultados_entrenamiento_inicial["horizonte_dias"] == horizonte_dias
    ].sort_values("f1_validacion", ascending=False)

    plt.plot(
        resultados_horizonte["modelo"],
        resultados_horizonte["f1_validacion"],
        marker="o",
        label=f"{horizonte_dias} días",
    )

plt.title("Comparación inicial de modelos por F1 en validación")
plt.xlabel("Modelo")
plt.ylabel("F1 en validación")
plt.xticks(rotation=45, ha="right")
plt.grid(True)
plt.legend()
plt.show()
```

## 23. Ajuste del umbral de decisión

Los modelos de clasificación generan una probabilidad estimada para la clase positiva. Por defecto, scikit-learn convierte esa probabilidad en clase usando un umbral de 0,5. Sin embargo, en este proyecto ese umbral no necesariamente es el más adecuado.

Para la aplicación final, la clase positiva representa una posible oportunidad de inversión. Por esta razón, no basta con maximizar la exactitud. Es especialmente importante controlar la precisión de la clase positiva, porque una recomendación de oportunidad debe ser relativamente confiable.

En esta sección se evalúan diferentes umbrales de decisión sobre el conjunto de validación. Esto permite analizar cómo cambian la precisión, el recall y el F1 cuando el modelo se vuelve más conservador o más flexible al emitir señales de oportunidad.

```python
def obtener_probabilidades_clase_positiva(modelo, x_datos):
    """
    Obtiene las probabilidades estimadas para la clase positiva.

    Esta función permite trabajar de forma uniforme con modelos de
    scikit-learn que implementan `predict_proba`.

    Args:
        modelo: Modelo de clasificación previamente entrenado.
        x_datos: Variables predictoras sobre las cuales se desea predecir.

    Returns:
        Arreglo de probabilidades asociadas a la clase positiva.
    """
    if not hasattr(modelo, "predict_proba"):
        return modelo.predict(x_datos)

    probabilidades = modelo.predict_proba(x_datos)
    clases_modelo = list(modelo.classes_)

    if 1 not in clases_modelo:
        return np.zeros(x_datos.shape[0])

    indice_clase_positiva = clases_modelo.index(1)

    return probabilidades[:, indice_clase_positiva]


def evaluar_modelo_con_umbral(y_real, probabilidades_clase_positiva, umbral_decision):
    """
    Evalúa un modelo binario usando un umbral de decisión específico.

    Args:
        y_real: Etiquetas reales.
        probabilidades_clase_positiva: Probabilidades estimadas para la clase 1.
        umbral_decision: Valor mínimo de probabilidad para clasificar una
            observación como oportunidad.

    Returns:
        Diccionario con métricas de clasificación y proporción de señales.
    """
    y_predicho = (probabilidades_clase_positiva >= umbral_decision).astype(int)

    cantidad_senales = int(y_predicho.sum())
    proporcion_senales = cantidad_senales / len(y_predicho)

    metricas_umbral = {
        "umbral_decision": umbral_decision,
        "accuracy": accuracy_score(y_real, y_predicho),
        "precision": precision_score(y_real, y_predicho, zero_division=0),
        "recall": recall_score(y_real, y_predicho, zero_division=0),
        "f1": f1_score(y_real, y_predicho, zero_division=0),
        "cantidad_senales": cantidad_senales,
        "proporcion_senales": proporcion_senales,
    }

    return metricas_umbral
```

```python
umbrales_decision = np.round(np.arange(0.30, 0.81, 0.025), 3)

resultados_umbrales = []

for horizonte_dias, modelos_horizonte in modelos_entrenados_iniciales.items():
    particiones = particiones_por_horizonte[horizonte_dias]

    x_validacion = particiones["x_validacion"]
    y_validacion = particiones["y_validacion"]

    for nombre_modelo, modelo in modelos_horizonte.items():
        if nombre_modelo == "dummy_mayoria":
            continue

        probabilidades_validacion = obtener_probabilidades_clase_positiva(
            modelo=modelo,
            x_datos=x_validacion,
        )

        for umbral_decision in umbrales_decision:
            metricas_umbral = evaluar_modelo_con_umbral(
                y_real=y_validacion,
                probabilidades_clase_positiva=probabilidades_validacion,
                umbral_decision=umbral_decision,
            )

            resultados_umbrales.append({
                "horizonte_dias": horizonte_dias,
                "modelo": nombre_modelo,
                **metricas_umbral,
            })

resultados_umbrales = pd.DataFrame(resultados_umbrales)

display(resultados_umbrales.head())
```

```python
mejores_umbrales_f1 = (
    resultados_umbrales
    .sort_values(
        by=["horizonte_dias", "f1"],
        ascending=[True, False],
    )
    .groupby("horizonte_dias")
    .head(10)
    .reset_index(drop=True)
)

display(mejores_umbrales_f1)
```

```python
resultados_umbrales_filtrados = resultados_umbrales[
    resultados_umbrales["proporcion_senales"] >= 0.05
].copy()

mejores_umbrales_precision = (
    resultados_umbrales_filtrados
    .sort_values(
        by=["horizonte_dias", "precision", "f1"],
        ascending=[True, False, False],
    )
    .groupby("horizonte_dias")
    .head(10)
    .reset_index(drop=True)
)

display(mejores_umbrales_precision)
```

```python
plt.figure(figsize=(12, 6))

for horizonte_dias in sorted(resultados_umbrales["horizonte_dias"].unique()):
    datos_horizonte = resultados_umbrales[
        (resultados_umbrales["horizonte_dias"] == horizonte_dias)
        & (resultados_umbrales["modelo"] == "gradient_boosting")
    ]

    plt.plot(
        datos_horizonte["recall"],
        datos_horizonte["precision"],
        marker="o",
        label=f"Gradient Boosting - {horizonte_dias} días",
    )

plt.title("Relación entre precisión y recall según el umbral de decisión")
plt.xlabel("Recall")
plt.ylabel("Precisión")
plt.grid(True)
plt.legend()
plt.show()
```

## Interpretación del ajuste de umbrales

El análisis de umbrales muestra que el umbral estándar de 0,5 no necesariamente es el más adecuado para este problema. Cuando se utilizan umbrales bajos, algunos modelos generan demasiadas señales de oportunidad, lo cual puede producir un F1 aparentemente competitivo pero poco útil para una aplicación de apoyo a decisiones de inversión.

En este contexto, una señal de oportunidad debe ser relativamente selectiva. No es deseable que el sistema recomiende oportunidad en la mayoría de los casos, porque perdería valor práctico para el usuario final. Por esta razón, además del F1, se analiza la precisión y la proporción de señales generadas.

Los resultados muestran que al aumentar el umbral de decisión, la precisión de la clase positiva mejora, aunque disminuye el recall. Este comportamiento es esperado: el modelo se vuelve más conservador, detecta menos oportunidades, pero las señales que emite tienden a ser más confiables.

El horizonte de 4 días presenta menor utilidad práctica, ya que para alcanzar una precisión aceptable se reduce demasiado la cantidad de señales. En cambio, los horizontes de 15 y 20 días muestran un mejor equilibrio para la aplicación, especialmente usando modelos como regresión logística o Gradient Boosting con umbrales de decisión más exigentes.

Con base en esta etapa, la aplicación final no debería interpretar cualquier probabilidad superior a 0,5 como una recomendación de inversión. En su lugar, se utilizará una capa de decisión conservadora que distinga entre señal fuerte, señal débil y ausencia de oportunidad clara.

## 24. Exploración de hiperparámetros

Después del entrenamiento inicial y del análisis de umbrales, se realiza una exploración controlada de hiperparámetros sobre los modelos más prometedores.

No se busca probar una cantidad excesiva de combinaciones, porque el tamaño del dataset es limitado y el problema financiero tiene una señal naturalmente ruidosa. En cambio, se evalúan configuraciones razonables para comparar si es posible mejorar el desempeño en validación sin aumentar demasiado el sobreajuste.

Los modelos seleccionados para esta etapa son:

- Regresión logística, por su estabilidad e interpretabilidad.
- Random Forest, por su capacidad de capturar relaciones no lineales.
- Gradient Boosting, por su buen desempeño inicial y su capacidad de construir ensambles secuenciales.

La comparación se realizará usando el conjunto de validación. Posteriormente, el mejor modelo seleccionado se evaluará una sola vez sobre el conjunto de prueba.

```python
from itertools import product
```

```python
def crear_combinaciones_hiperparametros(espacio_parametros):
    """
    Crea combinaciones de hiperparámetros a partir de un diccionario.

    Args:
        espacio_parametros: Diccionario donde cada llave representa un
            hiperparámetro y cada valor es una lista de posibles valores.

    Returns:
        Lista de diccionarios, donde cada diccionario contiene una combinación
        específica de hiperparámetros.
    """
    nombres_parametros = list(espacio_parametros.keys())
    valores_parametros = list(espacio_parametros.values())

    combinaciones = []

    for valores in product(*valores_parametros):
        combinacion = dict(zip(nombres_parametros, valores))
        combinaciones.append(combinacion)

    return combinaciones
```

```python
def crear_modelo_desde_configuracion(nombre_modelo, parametros):
    """
    Crea un modelo de clasificación a partir de una configuración específica.

    Args:
        nombre_modelo: Nombre del tipo de modelo a construir.
        parametros: Diccionario de hiperparámetros del modelo.

    Returns:
        Estimador de scikit-learn listo para entrenamiento.

    Raises:
        ValueError: Si el nombre del modelo no está soportado.
    """
    if nombre_modelo == "regresion_logistica":
        return Pipeline(
            steps=[
                ("escalador", StandardScaler()),
                (
                    "modelo",
                    LogisticRegression(
                        max_iter=3000,
                        random_state=42,
                        **parametros,
                    ),
                ),
            ]
        )

    if nombre_modelo == "random_forest":
        return RandomForestClassifier(
            random_state=42,
            n_jobs=-1,
            **parametros,
        )

    if nombre_modelo == "gradient_boosting":
        return GradientBoostingClassifier(
            random_state=42,
            **parametros,
        )

    raise ValueError(f"Modelo no soportado: {nombre_modelo}")
```

```python
espacios_hiperparametros = {
    "regresion_logistica": {
        "C": [0.01, 0.1, 1.0, 10.0],
        "class_weight": [None, "balanced"],
        "solver": ["lbfgs"],
    },
    "random_forest": {
        "n_estimators": [200, 400],
        "max_depth": [4, 6, 8],
        "min_samples_leaf": [10, 20, 40],
        "max_features": ["sqrt", "log2"],
    },
    "gradient_boosting": {
        "n_estimators": [100, 150, 250],
        "learning_rate": [0.03, 0.05, 0.08],
        "max_depth": [2, 3],
        "min_samples_leaf": [10, 20],
    },
}
```

```python
resultados_hiperparametros = []
modelos_hiperparametros = {}

for horizonte_dias, particiones in particiones_por_horizonte.items():
    modelos_hiperparametros[horizonte_dias] = {}

    x_entrenamiento = particiones["x_entrenamiento"]
    y_entrenamiento = particiones["y_entrenamiento"]
    x_validacion = particiones["x_validacion"]
    y_validacion = particiones["y_validacion"]

    for nombre_modelo, espacio_parametros in espacios_hiperparametros.items():
        combinaciones = crear_combinaciones_hiperparametros(espacio_parametros)

        for indice_configuracion, parametros in enumerate(combinaciones):
            modelo = crear_modelo_desde_configuracion(
                nombre_modelo=nombre_modelo,
                parametros=parametros,
            )

            modelo.fit(x_entrenamiento, y_entrenamiento)

            metricas_validacion = evaluar_modelo_clasificacion(
                modelo=modelo,
                x_datos=x_validacion,
                y_real=y_validacion,
            )

            probabilidades_validacion = obtener_probabilidades_clase_positiva(
                modelo=modelo,
                x_datos=x_validacion,
            )

            mejores_metricas_umbral = []

            for umbral_decision in umbrales_decision:
                metricas_umbral = evaluar_modelo_con_umbral(
                    y_real=y_validacion,
                    probabilidades_clase_positiva=probabilidades_validacion,
                    umbral_decision=umbral_decision,
                )

                metricas_umbral["modelo"] = nombre_modelo
                metricas_umbral["horizonte_dias"] = horizonte_dias
                metricas_umbral["parametros"] = parametros

                mejores_metricas_umbral.append(metricas_umbral)

            mejores_metricas_umbral = pd.DataFrame(mejores_metricas_umbral)

            candidatos_validos = mejores_metricas_umbral[
                mejores_metricas_umbral["proporcion_senales"].between(0.05, 0.25)
            ].copy()

            if candidatos_validos.empty:
                mejor_umbral = mejores_metricas_umbral.sort_values(
                    by="f1",
                    ascending=False,
                ).iloc[0]
            else:
                mejor_umbral = candidatos_validos.sort_values(
                    by=["precision", "f1"],
                    ascending=[False, False],
                ).iloc[0]

            clave_modelo = f"{nombre_modelo}_{indice_configuracion}"

            modelos_hiperparametros[horizonte_dias][clave_modelo] = modelo

            resultados_hiperparametros.append({
                "horizonte_dias": horizonte_dias,
                "modelo": nombre_modelo,
                "clave_modelo": clave_modelo,
                "parametros": parametros,
                "accuracy_validacion_05": metricas_validacion["accuracy"],
                "precision_validacion_05": metricas_validacion["precision"],
                "recall_validacion_05": metricas_validacion["recall"],
                "f1_validacion_05": metricas_validacion["f1"],
                "roc_auc_validacion": metricas_validacion["roc_auc"],
                "mejor_umbral": mejor_umbral["umbral_decision"],
                "precision_mejor_umbral": mejor_umbral["precision"],
                "recall_mejor_umbral": mejor_umbral["recall"],
                "f1_mejor_umbral": mejor_umbral["f1"],
                "proporcion_senales_mejor_umbral": mejor_umbral["proporcion_senales"],
            })

resultados_hiperparametros = pd.DataFrame(resultados_hiperparametros)

display(resultados_hiperparametros.head())
```

```python
mejores_modelos_validacion = (
    resultados_hiperparametros
    .sort_values(
        by=[
            "horizonte_dias",
            "precision_mejor_umbral",
            "f1_mejor_umbral",
            "roc_auc_validacion",
        ],
        ascending=[True, False, False, False],
    )
    .groupby("horizonte_dias")
    .head(10)
    .reset_index(drop=True)
)

display(mejores_modelos_validacion)
```

## Interpretación de la exploración de hiperparámetros

## 26. Evaluación final sobre el conjunto de prueba

Después de seleccionar los candidatos usando únicamente el conjunto de validación, se evalúa cada modelo seleccionado sobre el conjunto de prueba. Esta etapa permite estimar el desempeño del sistema sobre datos más recientes que no fueron usados ni para entrenamiento ni para selección de hiperparámetros.

Para esta evaluación se conserva el umbral de decisión seleccionado en validación. Esto evita ajustar el modelo directamente al conjunto de prueba y permite una evaluación más honesta de su capacidad de generalización.

```python
def evaluar_modelo_final_en_prueba(fila_candidato):
    """
    Evalúa un modelo candidato final sobre el conjunto de prueba.

    El modelo se toma desde las configuraciones entrenadas durante la búsqueda
    de hiperparámetros. El umbral de decisión utilizado es el seleccionado
    previamente sobre el conjunto de validación.

    Args:
        fila_candidato: Fila de DataFrame con la configuración seleccionada.

    Returns:
        Diccionario con métricas finales sobre el conjunto de prueba.
    """
    horizonte_dias = int(fila_candidato["horizonte_dias"])
    clave_modelo = fila_candidato["clave_modelo"]
    umbral_decision = float(fila_candidato["mejor_umbral"])

    modelo = modelos_hiperparametros[horizonte_dias][clave_modelo]

    particiones = particiones_por_horizonte[horizonte_dias]
    x_prueba = particiones["x_prueba"]
    y_prueba = particiones["y_prueba"]

    probabilidades_prueba = obtener_probabilidades_clase_positiva(
        modelo=modelo,
        x_datos=x_prueba,
    )

    metricas_prueba = evaluar_modelo_con_umbral(
        y_real=y_prueba,
        probabilidades_clase_positiva=probabilidades_prueba,
        umbral_decision=umbral_decision,
    )

    roc_auc_prueba = roc_auc_score(y_prueba, probabilidades_prueba)

    return {
        "horizonte_dias": horizonte_dias,
        "modelo": fila_candidato["modelo"],
        "clave_modelo": clave_modelo,
        "parametros": fila_candidato["parametros"],
        "umbral_decision": umbral_decision,
        "accuracy_prueba": metricas_prueba["accuracy"],
        "precision_prueba": metricas_prueba["precision"],
        "recall_prueba": metricas_prueba["recall"],
        "f1_prueba": metricas_prueba["f1"],
        "roc_auc_prueba": roc_auc_prueba,
        "cantidad_senales_prueba": metricas_prueba["cantidad_senales"],
        "proporcion_senales_prueba": metricas_prueba["proporcion_senales"],
    }


resultados_prueba_final = []

for _, fila_candidato in candidatos_finales_validacion.iterrows():
    resultados_prueba_final.append(
        evaluar_modelo_final_en_prueba(fila_candidato)
    )

resultados_prueba_final = pd.DataFrame(resultados_prueba_final)

display(resultados_prueba_final)
```

## Justificación del enfoque conservador

En el contexto de esta aplicación, se decidió priorizar la reducción de falsos positivos sobre el aumento del recall. Esta decisión se basa en la naturaleza del problema: una predicción positiva puede ser interpretada por el usuario como una señal de posible entrada al mercado.

Un falso positivo ocurre cuando el modelo indica que existe una oportunidad de inversión, pero posteriormente el retorno futuro no alcanza el umbral mínimo definido para el horizonte evaluado. Esto no necesariamente implica que el precio haya disminuido, pero sí significa que la oportunidad esperada no se materializó bajo los criterios del proyecto.

Por esta razón, se considera preferible emitir menos recomendaciones, pero más selectivas, en lugar de generar muchas señales con mayor riesgo de inducir al usuario a una compra poco favorable. En este sentido, el sistema se plantea como una herramienta de apoyo conservadora, no como un recomendador automático agresivo.

La principal desventaja de este enfoque es que el modelo puede dejar pasar oportunidades reales, lo cual se refleja en un recall bajo. Sin embargo, para el propósito de una aplicación orientada a usuarios finales, se considera más prudente reducir el riesgo de recomendaciones positivas incorrectas que maximizar la cantidad de oportunidades detectadas.

## Entrenamiento de modelos para despliegue

Una vez seleccionado el modelo final, se entrenan nuevamente los modelos de despliegue usando los conjuntos de entrenamiento y validación. El conjunto de prueba no se utiliza para este reentrenamiento, ya que fue reservado para la evaluación final del desempeño.

Esta estrategia permite aprovechar más datos históricos para el modelo que será usado por la aplicación, manteniendo al mismo tiempo una evaluación honesta sobre el conjunto de prueba realizada previamente.

```python
def entrenar_modelo_despliegue(fila_candidato):
    """
    Entrena un modelo final de despliegue usando entrenamiento y validación.

    El modelo se crea con los hiperparámetros seleccionados previamente y se
    ajusta con los datos de entrenamiento más validación. El conjunto de prueba
    no se usa en esta etapa.

    Args:
        fila_candidato: Fila del DataFrame de candidatos finales con el modelo,
            horizonte e hiperparámetros seleccionados.

    Returns:
        Modelo entrenado listo para guardarse y usarse en la aplicación.
    """
    horizonte_dias = int(fila_candidato["horizonte_dias"])
    nombre_modelo = fila_candidato["modelo"]
    parametros = fila_candidato["parametros"]

    modelo = crear_modelo_desde_configuracion(
        nombre_modelo=nombre_modelo,
        parametros=parametros,
    )

    particiones = particiones_por_horizonte[horizonte_dias]

    x_entrenamiento_completo = pd.concat(
        [
            particiones["x_entrenamiento"],
            particiones["x_validacion"],
        ],
        axis=0,
    )

    y_entrenamiento_completo = pd.concat(
        [
            particiones["y_entrenamiento"],
            particiones["y_validacion"],
        ],
        axis=0,
    )

    modelo.fit(x_entrenamiento_completo, y_entrenamiento_completo)

    return modelo


modelo_principal_despliegue = entrenar_modelo_despliegue(fila_modelo_principal)
modelo_apoyo_despliegue = entrenar_modelo_despliegue(fila_modelo_apoyo)

print("Modelos de despliegue entrenados correctamente.")
print(f"Modelo principal: {fila_modelo_principal['modelo']} - {int(fila_modelo_principal['horizonte_dias'])} días")
print(f"Modelo de apoyo: {fila_modelo_apoyo['modelo']} - {int(fila_modelo_apoyo['horizonte_dias'])} días")
```

## Importancia de variables del modelo principal

Para interpretar parcialmente el comportamiento del modelo principal, se revisa la importancia de las variables utilizadas por Gradient Boosting. Esta información permite identificar qué variables tuvieron mayor peso relativo en las divisiones construidas por el ensamble.

La importancia de variables no debe interpretarse como causalidad financiera. Sin embargo, sí ayuda a entender qué señales históricas fueron más utilizadas por el modelo para separar observaciones con y sin oportunidad de inversión.

```python
def obtener_estimador_base(modelo):
    """
    Extrae el estimador base de un modelo o pipeline.

    Args:
        modelo: Estimador de scikit-learn o Pipeline.

    Returns:
        Estimador final. Si el modelo es un Pipeline, retorna el paso llamado
        `modelo`; de lo contrario retorna el estimador original.
    """
    if hasattr(modelo, "named_steps") and "modelo" in modelo.named_steps:
        return modelo.named_steps["modelo"]

    return modelo


def obtener_importancia_variables(modelo, columnas_entrada):
    """
    Obtiene la importancia de variables para modelos basados en árboles.

    Args:
        modelo: Modelo entrenado.
        columnas_entrada: Lista de nombres de variables predictoras.

    Returns:
        DataFrame con variables e importancia relativa.
    """
    estimador_base = obtener_estimador_base(modelo)

    if not hasattr(estimador_base, "feature_importances_"):
        raise ValueError("El modelo seleccionado no expone feature_importances_.")

    importancia_variables = pd.DataFrame({
        "variable": columnas_entrada,
        "importancia": estimador_base.feature_importances_,
    })

    importancia_variables = importancia_variables.sort_values(
        "importancia",
        ascending=False,
    ).reset_index(drop=True)

    return importancia_variables


importancia_modelo_principal = obtener_importancia_variables(
    modelo=modelo_principal_despliegue,
    columnas_entrada=columnas_predictoras,
)

display(importancia_modelo_principal.head(15))
```

## Simulación de la recomendación para la aplicación

Antes de construir la aplicación final, se simula la salida esperada usando la observación más reciente del dataset. Esta prueba permite validar que el modelo puede recibir las variables predictoras, calcular probabilidades y traducirlas a una recomendación interpretable para el usuario.

La recomendación se genera a partir de una capa de decisión conservadora. Esta capa no reemplaza al modelo, sino que convierte sus probabilidades en un mensaje más comprensible para el usuario final.

```python
def generar_recomendacion_inversion(
    probabilidad_15d,
    probabilidad_20d,
    umbral_15d,
    umbral_20d,
):
    """
    Genera una recomendación interpretable a partir de probabilidades del modelo.

    La lógica de decisión prioriza señales conservadoras. El modelo principal
    corresponde al horizonte de 20 días y el modelo de 15 días se utiliza como
    apoyo para enriquecer la interpretación.

    Args:
        probabilidad_15d: Probabilidad estimada de oportunidad a 15 días.
        probabilidad_20d: Probabilidad estimada de oportunidad a 20 días.
        umbral_15d: Umbral de decisión del modelo de 15 días.
        umbral_20d: Umbral de decisión del modelo de 20 días.

    Returns:
        Diccionario con nivel de señal, recomendación e interpretación.
    """
    zona_monitoreo_15d = umbral_15d * 0.85
    zona_monitoreo_20d = umbral_20d * 0.85

    if probabilidad_20d >= umbral_20d and probabilidad_15d >= zona_monitoreo_15d:
        return {
            "nivel_senal": "Oportunidad fuerte",
            "recomendacion": "El sistema identifica una señal conservadora favorable.",
            "interpretacion": (
                "El modelo principal de 20 días supera el umbral de decisión y "
                "el modelo de 15 días acompaña con una probabilidad relevante. "
                "Esto sugiere una posible oportunidad de inversión bajo los "
                "criterios definidos en el proyecto."
            ),
        }

    if probabilidad_20d >= umbral_20d:
        return {
            "nivel_senal": "Oportunidad potencial a 20 días",
            "recomendacion": "Podría existir una oportunidad en el horizonte mensual.",
            "interpretacion": (
                "El modelo principal supera el umbral conservador de 20 días, "
                "pero la señal de 15 días no acompaña con suficiente fuerza. "
                "Se recomienda interpretar el resultado con moderación."
            ),
        }

    if probabilidad_15d >= umbral_15d:
        return {
            "nivel_senal": "Oportunidad potencial a 15 días",
            "recomendacion": "Existe una señal de corto plazo, pero no es confirmada por el horizonte principal.",
            "interpretacion": (
                "El modelo de 15 días supera su umbral, pero el modelo principal "
                "de 20 días no confirma una señal fuerte. El sistema sugiere "
                "monitorear antes de tomar una decisión."
            ),
        }

    if probabilidad_20d >= zona_monitoreo_20d or probabilidad_15d >= zona_monitoreo_15d:
        return {
            "nivel_senal": "Monitorear",
            "recomendacion": "No hay señal fuerte, pero conviene revisar datos actualizados próximamente.",
            "interpretacion": (
                "Las probabilidades estimadas se encuentran en una zona intermedia. "
                "El sistema no emite una recomendación positiva fuerte, pero sugiere "
                "monitorear la evolución del ETF con datos más recientes."
            ),
        }

    return {
        "nivel_senal": "Sin oportunidad clara",
        "recomendacion": "El sistema no identifica una oportunidad conservadora en este momento.",
        "interpretacion": (
            "Las probabilidades estimadas no superan los umbrales definidos. "
            "Bajo el enfoque conservador del proyecto, es preferible no emitir "
            "una señal positiva de inversión."
        ),
    }
```

```python
ultima_observacion = datos_features.dropna(subset=columnas_predictoras).tail(1).copy()

x_ultima_observacion = ultima_observacion[columnas_predictoras]

probabilidad_20d = obtener_probabilidades_clase_positiva(
    modelo=modelo_principal_despliegue,
    x_datos=x_ultima_observacion,
)[0]

probabilidad_15d = obtener_probabilidades_clase_positiva(
    modelo=modelo_apoyo_despliegue,
    x_datos=x_ultima_observacion,
)[0]

umbral_20d = float(fila_modelo_principal["mejor_umbral"])
umbral_15d = float(fila_modelo_apoyo["mejor_umbral"])

recomendacion_simulada = generar_recomendacion_inversion(
    probabilidad_15d=probabilidad_15d,
    probabilidad_20d=probabilidad_20d,
    umbral_15d=umbral_15d,
    umbral_20d=umbral_20d,
)

resultado_simulacion_app = {
    "ultima_fecha_disponible": ultima_observacion["fecha"].iloc[0],
    "ultimo_precio_cierre": ultima_observacion["precio_cierre"].iloc[0],
    "probabilidad_15d": probabilidad_15d,
    "umbral_15d": umbral_15d,
    "probabilidad_20d": probabilidad_20d,
    "umbral_20d": umbral_20d,
    **recomendacion_simulada,
}

display(pd.DataFrame([resultado_simulacion_app]))
```

```python
import json
import joblib


def convertir_a_json_serializable(valor):
    """
    Convierte objetos comunes de pandas y numpy a formatos compatibles con JSON.

    Args:
        valor: Objeto que se desea serializar.

    Returns:
        Objeto equivalente compatible con `json.dump`.
    """
    if isinstance(valor, dict):
        return {
            clave: convertir_a_json_serializable(valor_dict)
            for clave, valor_dict in valor.items()
        }

    if isinstance(valor, list):
        return [convertir_a_json_serializable(elemento) for elemento in valor]

    if isinstance(valor, tuple):
        return [convertir_a_json_serializable(elemento) for elemento in valor]

    if isinstance(valor, np.integer):
        return int(valor)

    if isinstance(valor, np.floating):
        return float(valor)

    if isinstance(valor, pd.Timestamp):
        return str(valor)

    return valor


ruta_modelo_principal = ruta_modelos / "modelo_principal_gradient_boosting_20d.joblib"
ruta_modelo_apoyo = ruta_modelos / "modelo_apoyo_gradient_boosting_15d.joblib"
ruta_metadata_modelos = ruta_modelos / "metadata_modelos_finales.json"

joblib.dump(modelo_principal_despliegue, ruta_modelo_principal)
joblib.dump(modelo_apoyo_despliegue, ruta_modelo_apoyo)

metadata_modelos_finales = {
    "descripcion": "Modelos finales para sistema conservador de apoyo a inversión en ETF VOO.",
    "modelo_principal": {
        "archivo": ruta_modelo_principal.name,
        "modelo": fila_modelo_principal["modelo"],
        "horizonte_dias": int(fila_modelo_principal["horizonte_dias"]),
        "umbral_decision": float(fila_modelo_principal["mejor_umbral"]),
        "umbral_retorno_objetivo": float(configuracion_objetivos[int(fila_modelo_principal["horizonte_dias"])]),
        "parametros": fila_modelo_principal["parametros"],
    },
    "modelo_apoyo": {
        "archivo": ruta_modelo_apoyo.name,
        "modelo": fila_modelo_apoyo["modelo"],
        "horizonte_dias": int(fila_modelo_apoyo["horizonte_dias"]),
        "umbral_decision": float(fila_modelo_apoyo["mejor_umbral"]),
        "umbral_retorno_objetivo": float(configuracion_objetivos[int(fila_modelo_apoyo["horizonte_dias"])]),
        "parametros": fila_modelo_apoyo["parametros"],
    },
    "columnas_predictoras": columnas_predictoras,
    "rango_fechas_dataset": {
        "fecha_minima": str(datos_voo["fecha"].min().date()),
        "fecha_maxima": str(datos_voo["fecha"].max().date()),
    },
    "politica_decision": {
        "enfoque": "conservador",
        "prioridad": "reducir falsos positivos en señales positivas de inversión",
        "advertencia": (
            "Sistema académico de apoyo a decisión. No constituye asesoría financiera."
        ),
    },
}

metadata_modelos_finales = convertir_a_json_serializable(metadata_modelos_finales)

with open(ruta_metadata_modelos, "w", encoding="utf-8") as archivo_metadata:
    json.dump(
        metadata_modelos_finales,
        archivo_metadata,
        ensure_ascii=False,
        indent=4,
    )

print("Artefactos guardados:")
print(ruta_modelo_principal)
print(ruta_modelo_apoyo)
print(ruta_metadata_modelos)
```

## Conclusiones del notebook

En este notebook se desarrolló el proceso completo de exploración, preparación, entrenamiento y selección de modelos para apoyar decisiones de inversión sobre el ETF VOO.

Primero se realizó una limpieza del dataset original, transformando fechas, precios, volumen y porcentajes a formatos adecuados para el análisis. También se eliminaron registros sin operación reportada, ya que no representaban días normales de negociación.

Posteriormente se formularon variables objetivo para tres horizontes temporales: 4, 15 y 20 días. Cada objetivo se definió como una clasificación binaria basada en si el retorno futuro superaba un umbral mínimo. Esta formulación permitió transformar el problema financiero en una tarea de aprendizaje supervisado.

Luego se construyeron variables predictoras basadas en retornos históricos, volatilidad, momentum, distancia frente a medias móviles, distancia frente a máximos y mínimos recientes, y comportamiento relativo del volumen. Todas las variables fueron calculadas usando únicamente información disponible hasta la fecha de cada observación, evitando fuga de información.

Se entrenaron y compararon diferentes modelos: un modelo ingenuo de referencia, regresión logística, Random Forest, Gradient Boosting y una red neuronal multicapa. Los resultados iniciales mostraron que existe una señal predictiva moderada, pero también evidenciaron que el problema es ruidoso y que el uso del umbral estándar de 0,5 no es suficiente para construir una recomendación útil.

La exploración de umbrales permitió construir un sistema más conservador. En lugar de maximizar únicamente el recall, se decidió priorizar la precisión de la clase positiva, debido a que una señal positiva puede inducir al usuario a tomar una decisión de inversión. Esta decisión reduce falsos positivos, aunque implica dejar pasar oportunidades reales.

El modelo seleccionado como componente principal fue Gradient Boosting para el horizonte de 20 días. Este modelo presentó el mejor comportamiento global en prueba, especialmente en ROC-AUC, y es coherente con un horizonte mensual de inversión. Como apoyo, se conserva un modelo Gradient Boosting para 15 días.

La principal limitación del sistema es su baja cobertura: el modelo emite pocas señales positivas. Por lo tanto, la aplicación final no debe presentarse como un sistema automático de compra o venta, sino como una herramienta conservadora de apoyo a la decisión que permite identificar posibles oportunidades bajo criterios definidos.
